```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/monthly_neighbourhood_safety_from_cleaned.csv")

plt.figure(figsize=(8, 5))
plt.boxplot(df["safety_index"].dropna(), vert=True, patch_artist=True)
plt.ylabel("Safety Index")
plt.title("Box Plot of Safety Index")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```
